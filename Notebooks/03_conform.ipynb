```python
import pandas as pd
from pathlib import Path

transform_file = Path("../data/transform/fuel_prices.csv")
conform_dir = Path("../data/conform")
conform_dir.mkdir(parents=True, exist_ok=True)

prices = pd.read_csv(transform_file, parse_dates=["date"])

long = prices.melt(id_vars="date", var_name="fuel_type", value_name="price_rm")
long = long.dropna(subset=["price_rm"])

long["fuel_type"] = long["fuel_type"].str.upper().str.replace("_", " ")

long.to_csv(conform_dir / "fuel_prices_long.csv", index=False)
print(len(long), "rows saved")
print(long["fuel_type"].unique())
```
