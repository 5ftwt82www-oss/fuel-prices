```python
import pandas as pd
from pathlib import Path

raw_dir = Path("../data/raw/fuel_prices")
transform_dir = Path("../data/transform")
transform_dir.mkdir(parents=True, exist_ok=True)

latest_file = sorted(raw_dir.glob("fuelprice_*.csv"))[-1]
print("Reading:", latest_file.name)

df = pd.read_csv(latest_file)

prices = df[df["series_type"] == "level"].copy()
prices["date"] = pd.to_datetime(prices["date"])
prices = prices.sort_values("date")

print(len(df), "rows in raw")
print(len(prices), "price rows kept")
print(prices.dtypes)
```

```python
prices = prices.drop(columns=["series_type"])
prices = prices.drop_duplicates(subset="date", keep="last")

print(prices.isna().sum())

out_file = transform_dir / "fuel_prices.csv"
prices.to_csv(out_file, index=False)
print("Saved", len(prices), "rows to", out_file)
```
